```python
# IT2-3-1.Les genres botaniques représentés à Grenoble : identifier et compter
##Quel est le genre botanique le plus représenté ?
n_genre = {}
for ligne in data[1:]:
    genre = ligne[13]
    if genre != "":
        if genre in n_genre:
            n_genre[genre] += 1
        else:
            n_genre[genre] = 1
print(n_genre)
##Initialiser les variables
genre_max = ""    ## crée une variable vide qui servira à mémoriser le nom du genre le plus représenté.
nombre_max = 0    ##initialise le nombre maximal d'arbres à zéro.
for genre in n_genre:    #Parcourir les genres  une par une
    if n_genre[genre] > nombre_max:    
        nombre_max = n_genre[genre]
        genre_max = genre
print("Le genre le plus représenté est", genre_max)
print("Nombre de représentants :", nombre_max) 
```

```python
#### IT2-3-8. Essai_Graphe:tracer l'histogramme représentant le nombre d'arbres plantés par année.
import matplotlib.pyplot as plt
# Trier les années dans l'ordre chronologique
annees = sorted(arbres_annee.keys(), key=int)

# Récupérer les nombres pour chaque année
nb_arbres = [arbres_annee[annee] for annee in annees]
nb_genres = [len(genres_annee[annee]) for annee in annees]

# Créer le graphique
fig, ax1 = plt.subplots(figsize=(12, 6))

# Barres : nombre d'arbres
ax1.bar(annees, nb_arbres, color="skyblue", label="Nombre d'arbres")
ax1.set_xlabel("Année")
ax1.set_ylabel("Nombre d'arbres", color="blue")
ax1.tick_params(axis="y", labelcolor="blue")
ax1.tick_params(axis="x", rotation=45)

# Deuxième axe : nombre de genres
ax2 = ax1.twinx()
ax2.plot(annees, nb_genres, color="red", marker="o",
         label="Nombre de genres")
ax2.set_ylabel("Nombre de genres botaniques", color="red")
ax2.tick_params(axis="y", labelcolor="red")

# Titre et légende
plt.title("Évolution du nombre d'arbres et de genres botaniques par année")
fig.tight_layout()
plt.show()
```

```python
#methode Haversine:
import math

def distance_haversine(arbre1, arbre2):
    lat1, lon1 = arbre1
    lat2, lon2 = arbre2

    # Conversion des degrés en radians
    lat1 = math.radians(lat1)
    lon1 = math.radians(lon1)
    lat2 = math.radians(lat2)
    lon2 = math.radians(lon2)

    # Différences entre les coordonnées
    dlat = lat2 - lat1
    dlon = lon2 - lon1

    # Formule de Haversine
    a = (math.sin(dlat / 2) ** 2
         + math.cos(lat1) * math.cos(lat2)
         * math.sin(dlon / 2) ** 2)

    c = 2 * math.asin(math.sqrt(a))

    # Rayon moyen de la Terre en mètres
    rayon_terre = 6371000
    return rayon_terre * c


arbre1 = (45.1885, 5.7245)
arbre2 = (45.1900, 5.7300)

distance = distance_haversine(arbre1, arbre2)

print("Distance entre les deux arbres :", round(distance, 2), "mètres")
```

```python
#Quel est l'arbre le plus solitaire ?
import json
import numpy as np
from sklearn.neighbors import BallTree

# Garder les arbres qui possèdent des coordonnées
arbres_valides = arbres.dropna(subset=["GeoJSON"]).copy()

# Extraire les coordonnées GPS
coordonnees = [json.loads(x)["coordinates"]for x in arbres_valides["GeoJSON"]]

# Convertir en radians : latitude puis longitude
points = np.radians( [[c[1], c[0]] for c in coordonnees])

# Rechercher les deux arbres les plus proches
tree = BallTree(points, metric="haversine")
distances, indices = tree.query(points, k=2)

# Calculer la distance au voisin le plus proche en mètres
distances_metres = distances[:, 1] * 6371000

# Trouver l'arbre le plus solitaire
indice = np.argmax(distances_metres)

print("Arbre le plus solitaire")
print("Distance au voisin le plus proche :",
      round(distances_metres[indice], 2), "mètres")
print("Identifiant :",
      arbres_valides.iloc[indice]["ELEM_POINT_ID"])
print("Genre botanique :",
      arbres_valides.iloc[indice]["GENRE_BOTA"])
print("Espèce :",
      arbres_valides.iloc[indice]["ESPECE"])
print("Coordonnées GPS :", coordonnees[indice])
```
